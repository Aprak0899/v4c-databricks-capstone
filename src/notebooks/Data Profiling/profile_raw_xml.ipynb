## 2. Structural validation (strict XML) — binary: valid / invalid

```python
import xml.etree.ElementTree as ET

def is_valid_xml(file_path: str) -> bool:
    try:
        ET.parse(file_path)
        return True
    except ET.ParseError as e:
        print(f"XML parse error: {e}")
        return False
    
assert is_valid_xml("/Volumes/dev_automotive/landing/landing_raw/final_geografic.xml"), \
       "❌ Malformed XML file"
print("✅ XML is well-formed")
```
