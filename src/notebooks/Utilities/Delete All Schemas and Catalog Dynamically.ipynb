```python
# Create a text widget for catalog name
dbutils.widgets.text("catalog_name", "default_catalog", "Catalog Name")

# Get the catalog name from the widget
catalog_name = dbutils.widgets.get("catalog_name")

```

```python
# 1. Drop all user schemas
schemas = spark.sql(f"""
SELECT schema_name
FROM {catalog_name}.information_schema.schemata
WHERE schema_name NOT IN ('information_schema')
""").collect()

for row in schemas:
    spark.sql(f"DROP SCHEMA IF EXISTS `{catalog_name}`.`{row.schema_name}` CASCADE")

# 2. Drop the catalog
spark.sql(f"DROP CATALOG IF EXISTS `{catalog_name}`")
```
